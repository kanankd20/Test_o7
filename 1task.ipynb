```python
import pandas as pd
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
#data undersanding 
pd=pd.read_csv('Nike_Sales_Uncleaned.csv')
pd.head()
```

```python
pd.info()
```

```python
pd.describe()
```

```python
#data cleaning 
pd["Discount_Applied"].isnull().sum()
```

```python
pd["Discount_Applied"].fillna(0, inplace=True)
```

```python
pd['Order_Date'] = pd['Order_Date'].astype('datetime64[ns]')
pd['Order_Date'].fillna(pd['Order_Date'].mode()[0], inplace=True)
```

```python
pd['Size'].value_counts()
```

```python
pd['Size'].fillna(pd['Size'].mode()[0],inplace=True)
```

```python
pd.info()
```

```python
pd.isnull().sum()
```

```python
pd['Units_Sold'].fillna(pd['Units_Sold'].mean(), inplace=True)
pd['MRP'].fillna(pd['MRP'].mean(), inplace=True)
```

```python
pd.isnull().sum()
```

```python
pd.describe()
```

```python
#data analysis 
#determine total units sold , total revenue and total profit 
total_units_sold = pd['Units_Sold'].sum()
total_revenue = (pd['Units_Sold'] * pd['MRP']).sum()
total_profit = (pd['Units_Sold'] * (pd['MRP'] - pd['Cost_Price'])).sum()
print("Total Units sold ", total_units_sold)
print("Total Revenue ", total_revenue)
print("Total Profit ", total_profit)
```

```python
pd.info()
```

```python
#analyse sales across different product lines 
product_lines_sales= pd.groupby('Product_Line')['Units_Sold'].sum().sort_values(ascending=True)
#sales across regions to analyse performance 
sales_regiosn= pd.groupby('Region')['Units_Sold'].sum().sort_values(ascending=False)
 
```

```python
pd['Sales_Channel'].value_counts()
```

```python
#compare online and retail sales performance 
online_sales = pd[pd['Sales_Channel'] == 'Online']['Units_Sold'].sum()
retail_sales = pd[pd['Sales_Channel'] == 'Retail']['Units_Sold'].sum()
```

```python
pd['Gender_Category'].value_counts()
```

```python
#gender category with max.sales
max_sales_gender = pd.groupby('Gender_Category')['Units_Sold'].sum().idxmax()
print(max_sales_gender)
```

```python
#data visualisation 
#product lines with revenue 
product_lines_revenue = pd.groupby('Product_Line')['Revenue'].sum().sort_values(ascending=False)
plt.figure(figsize=(10,6))
sns.barplot(x=product_lines_revenue.index, y=product_lines_revenue.values)
plt.xlabel('Product Line')
plt.ylabel('Revenue')

```
